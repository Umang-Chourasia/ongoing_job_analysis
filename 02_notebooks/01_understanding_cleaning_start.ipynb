```python
rows_without_company_name = df[df['company'].isna()]
rows_without_company_name
```

```python
df['company'] = (df['company']
                .astype(str)
                .str.strip()
                .str.replace(r'\s+', ' ', regex = True)
                .str.lower()
                )

company_map = {
    "tata consultancy services": "tcs",
    "tata consultancy service": "tcs",
    "tcs": "tcs",
    "t.c.s.": "tcs",
    "infosys limited": "infosys",
    "google llc": "google",
    'sankhyana consultancy services':'scs'
}

df['company'] = df['company'].replace(company_map)
```

```python
keywords = ['hiring', 'client', 'firm', 'leading', 'mnc', 'not available']
max_words = 6

def clean_company_name(name):
    if pd.isna(name) or any(word in name.lower() for word in keywords) or len(name.split()) > max_words:
        return "unknown company"
    return name

df['company'] = df['company'].apply(clean_company_name)
print(df['company'])
```

```python
df['job_tittle'] = (
    df['job_detail']
      .str.split("Role:").str[1]
      .str.split("',").str[0]
      .str.split('-').str[0]     # removes "- other"
      .str.split('(').str[0]
      .str.split('/').str[0]
      .str.strip()
)
```

```python
role_mapping = {
    "junior data analyst": "data analyst",
    "senior data analyst": "data analyst",
    "network security engineer": "network engineer",
    "business development executive": "business development",
    "junior system analyst": "systems analyst",
    "senior system analyst": "systems analyst",
    "technical support": "it support",
    "treasury analyst": "financial analyst",
    "operations manager": "operations",
    "assistant manager": "manager",
    "chief accountant": "accountant",
    "automation test engineer": "automation engineer",
    "it project lead": "it project manager",
    "private attorney": "attorney",
    "customer success manager": "customer success",
    "audit executive": "audit manager",
    "quality assurance and testing": "quality assurance"
}
```

```python
df['job_tittle'] = df['job_tittle'].str.lower()
df['job_tittle'] = df['job_tittle'].replace(role_mapping)
```
