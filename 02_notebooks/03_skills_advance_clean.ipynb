```python
import pandas as pd
import numpy as np
import ast
import re
```

```python
df = pd.read_csv("company_clean_2.0.csv")
```

```python
df.rename(columns={'required_skill': 'skills'}, inplace=True)
```

```python
df['skills'] = df['skills'].apply(lambda x:x.lower())
```

```python
df['skills'] = df['skills'].apply(lambda x:x.strip())
```

```python
skill_map = {
    # 🔹 Programming / Languages
    'python': 'python',

    # 🔹 SQL & Databases
    'mysql': 'sql',
    'postgresql': 'sql',
    'sql server': 'sql',
    'oracle sql': 'sql',
    'sql database': 'sql',
    'sql': 'sql',

    # 🔹 Excel / BI
    'excel': 'excel',
    'advanced excel': 'excel',
    'ms excel': 'excel',

    'power bi': 'power bi',
    'powerbi': 'power bi',
    'tableau': 'tableau',

    # 🔹 Data / ML
    'machine learning': 'machine learning',
    'artificial intelligence': 'machine learning',
    'deep learning': 'machine learning',

    'statistics': 'statistics',
    'statistical analysis': 'statistics',

    # 🔹 Libraries
    'pandas': 'pandas',
    'numpy': 'numpy',

    # 🔹 Big Data
    'spark': 'spark',
    'pyspark': 'spark',
    'hadoop': 'big data',

    # 🔹 Cloud
    'aws': 'aws',
    'amazon web services': 'aws',
    'azure': 'cloud',
    'microsoft azure': 'cloud',
    'gcp': 'cloud',

    # 🔹 Visualization / Analytics
    'data visualization': 'power bi',
    'business intelligence': 'bi',

    # 🔹 DevOps / Tools
    'docker': 'devops',
    'kubernetes': 'devops',

    # 🔹 Security (you have a LOT of this)
    'cyber security': 'security',
    'information security': 'security',
    'network security': 'security',
    'endpoint security': 'security',
    'cloud security': 'security',

    # 🔹 Networking
    'networking': 'networking',
    'tcp': 'networking',
    'dns': 'networking'
}
```

```python
import ast

# Step 1: convert string → list
df['skills'] = df['skills'].apply(ast.literal_eval)

# Step 2: clean + normalize + map
df['skills'] = df['skills'].apply(
    lambda x: list({
        skill_map.get(s.strip().lower(), s.strip().lower())
        for s in x
    })
)
```

```python
# type(df['skills'].iloc[0])
```

```python
skills = [
    "python",
    "sql",
    "excel",
    "power bi",
    "tableau",
    "machine learning",
    "pandas",
    "numpy",
    "spark",
    "aws",
    "statistics"
]
```

```python
for skill in skills:
    df[f"has_{skill.replace(' ', '_')}"] = df["skills"].apply(
        lambda x: 1 if skill in x else 0
    )
```
